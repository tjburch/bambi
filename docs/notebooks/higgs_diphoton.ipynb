# Finding a particle: a Higgs diphoton case study

This case study applies Bambi's [nonlinear-formula API](nonlinear_expressions.ipynb)
to a scientific-style analysis of a particle resonance. We build a signal-plus-background
model, visualize its fitted components, and examine its assumptions and model evidence.

An unstable particle can leave a localized excess, or *resonance*, in the invariant mass of its
decay products. Invariant mass combines their measured energies and momenta into an estimate
of the parent particle's mass. The Higgs discovery is a famous example: pairs of photons
showed a small excess near 125 GeV on top of a much larger, smoothly falling background.

We use the **diphoton channel only**; the discovery combined evidence from several decay
channels. Here we use real CMS discovery data to illustrate **composition of nonlinear components**.
The expected count is `background(mass) + resonance(mass)`, with an unknown resonance location,
width, and amplitude. The response is a count, and its mean is an additive combination of two positive physical
components.

## Data and the role of weights

The [CMS public discovery-model release](https://doi.org/10.17181/c2948-e8875), licensed
**CC BY 4.0** by the CMS Collaboration, supplies observed diphoton data in 11 categories at
collision energies of 7 and 8 TeV. Categories distinguish events with different mass resolutions
and signal-to-background ratios. Our CSV contains real, unweighted counts in 1 GeV bins.
We fit the full 100–180 GeV range used in the analysis, and zoom the fitted spectrum
to 110–150 GeV as in the
[discovery paper's Figure 3](https://cms-results.web.cern.ch/cms-results/public-results/publications/HIG-12-028/index.html).
A second CSV supplies category information from its Table 2.
[Full provenance, category mapping, and extraction instructions](data/higgs_diphoton_README.md)
accompany the data.

The discovery plot used category weights proportional to $S/(S+B)$ to make high-purity
categories more visible. Such a weighted sum is **not itself Poisson**. We retain the category
counts in a simultaneous Poisson fit, preserving their different sensitivities, and apply
weights when displaying the combined spectrum. Bambi's `weighted()` response multiplies
log-likelihood contributions; that is a different operation.

We retain the analysis's category structure, with these deliberate simplifications:

| Ingredient | This example | CMS discovery analysis |
| --- | --- | --- |
| Signal shape | One Gaussian per category, with a shared mass and width scale | Calibrated Gaussian mixtures, including different vertex-reconstruction components |
| Background | A quadratic log-rate per category over 100–180 GeV | Category-dependent degree 3–5 polynomials over 100–180 GeV, selected through signal-bias studies |
| Uncertainty and inference | Bayesian parameter estimation with fixed relative category yields and resolutions | Likelihood-based significance and interval calculations with correlated experimental and theoretical nuisance parameters |

Our goal is to estimate the known-region excess and understand how its components combine.
The positive amplitude prior excludes an exactly background-only model, so an amplitude interval
above zero is not a discovery test. The intervals also omit experimental systematic uncertainty.
We will compare with a separately fitted background-only model and check sensitivity
to removing background curvature.

```python
# | code-fold: true
%config InlineBackend.figure_format = "retina"
%config InlineBackend.print_figure_kwargs = {'bbox_inches': None}

import arviz as az
import bambi as bmb
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

plt.style.use("bambi.mplstyle")
plot_colors = plt.rcParams["axes.prop_cycle"].by_key()["color"]
```

```python
higgs_categories = pd.read_csv("data/higgs_diphoton_categories.csv").set_index(
    "category"
)
higgs = pd.read_csv("data/higgs_diphoton_counts.csv")
higgs["category"] = pd.Categorical(higgs["category"], categories=higgs_categories.index)
higgs["x"] = (higgs["mass"] - 125) / 20

# Fixed relative peak heights and resolutions from the published category information.
peak_heights = higgs_categories["signal_events"] / higgs_categories["sigma_eff"]
higgs_categories["peak_fraction"] = peak_heights / peak_heights.sum()
higgs_categories["resolution_ratio"] = higgs_categories["sigma_eff"] / 2
for column in ["peak_fraction", "resolution_ratio"]:
    higgs[column] = higgs["category"].map(higgs_categories[column]).astype(float)

observed_counts = higgs.pivot(index="mass", columns="category", values="counts")
plt.stairs(observed_counts.sum(axis=1), np.arange(100, 181), color="black")
plt.xlabel("Diphoton invariant mass (GeV)")
plt.ylabel("Events / 1 GeV")
plt.title("CMS discovery data: all categories combined")
plt.show()
```

## A shared resonance in several categories

For mass bin $i$ and category $c$, we model

$$
N_{ic} \sim \operatorname{Poisson}(\lambda_{ic}), \qquad
\lambda_{ic} = \underbrace{\exp(b_{0c}+b_{1c}x_i+b_{2c}x_i^2)}_{B_{ic}}
+ \underbrace{A p_c \exp\!\left[-\frac12
\left(\frac{m_i-m_H}{\sigma r_c}\right)^2\right]}_{S_{ic}},
\qquad x_i = \frac{m_i-125}{20}.
$$

Here $p_c$ is the fixed `peak_fraction` (summing to one), and $r_c$ is the fixed
`resolution_ratio`. We construct $p_c$ proportional to the published signal yield divided by
its resolution, so the relative *integrated* signal yields also match the table. Every category
has its own background intercept, slope, and curvature, while $m_H$, $A$, and $\sigma$ are shared.

The published $\sigma_{\mathrm{eff}}$ is half the width of the narrowest interval containing
68% of the signal. For a non-Gaussian signal it is not a standard deviation. Using its category
ratios as Gaussian standard-deviation ratios is a useful approximation here.

$A=\exp(\mathrm{log\_amp})$ is the **combined peak height**, in expected signal events per
1 GeV bin at $m_H$, not the total signal yield. $\sigma=\exp(\mathrm{log\_sigma})$ is a reference
width; each category's width is $\sigma r_c$. These widths describe detector resolution, not
the Higgs boson's intrinsic decay width. The positive parameterization avoids negative rates
or widths. Evaluating the curve at bin centers is a midpoint approximation for these fixed,
equal-width bins; changing the bin width would require changing the rate normalization.

For Poisson's log link the nonlinear parent must be $\eta_{ic}=\log(B_{ic}+S_{ic})$.
Bambi applies the inverse link once to obtain $\lambda_{ic}$. Omitting the outer `log` would
exponentiate the expected counts again.

`nlpars` declares the six nonlinear parameter names. The `0 + category` formulas give
`b0`, `b1`, and `b2` separate coefficients for each category. The other three parameters default to
intercept-only formulas, equivalent to writing `log_amp ~ 1`, `m_h ~ 1`, and `log_sigma ~ 1`.

```python
higgs_formula = bmb.Formula(
    "counts ~ log(exp(b0 + b1 * x + b2 * x**2) + "
    "peak_fraction * exp(log_amp - 0.5 * "
    "((mass - m_h) / (exp(log_sigma) * resolution_ratio))**2))",
    "b0 ~ 0 + category",
    "b1 ~ 0 + category",
    "b2 ~ 0 + category",
    nlpars=("b0", "b1", "b2", "log_amp", "m_h", "log_sigma"),
)
```

Nonlinear coefficients are not automatically scaled to the response, so we set explicit
priors. Normal$(125,3)$ for $m_H$ covers roughly 119–131 GeV with 95% probability.
Normal$(\log 2,0.3)$ for `log_sigma` allows a reference width of roughly 1.1–3.6 GeV,
and Normal$(\log 40,0.8)$ for `log_amp` allows peak heights of roughly 8–190 events per bin.
These priors describe a detector-scale resonance while leaving its location and size to be
estimated from the data. They describe a retrospective, targeted example, not the prior
knowledge of a historical search across all possible Higgs masses.

For every background intercept we use Normal$(\log 30,2)$: at 125 GeV its prior median is
30 events per bin, with a 95% interval of roughly 0.6–1,500. This broad, proper prior allows
large category differences without centering each prior on a background estimate from these
data. Both competing models will use exactly the same background priors, which matters for
the later evidence comparison. We use the published background estimates only to initialize
the sampler; starting values are not prior information.

Normal$(-0.5,0.5)$ for the slope favors a falling spectrum but permits increasing backgrounds.
Scaling mass by 20 GeV makes this prior easy to interpret. We allow curvature with
Normal$(0,0.5)$ for each `b2`. At $|x|=1$, one standard deviation changes the background by a
factor of about 1.65. The wider fit range helps constrain the background, but also makes a
single exponential too restrictive here. A quadratic log-rate adds flexibility while keeping
rates positive; it is still simpler than the original category-dependent polynomials and their
signal-bias validation.

```python
higgs_priors = {
    "b0": {"category": bmb.Prior("Normal", mu=np.log(30), sigma=2)},
    "b1": {"category": bmb.Prior("Normal", mu=-0.5, sigma=0.5)},
    "b2": {"category": bmb.Prior("Normal", mu=0, sigma=0.5)},
    "log_amp": {"Intercept": bmb.Prior("Normal", mu=np.log(40), sigma=0.8)},
    "m_h": {"Intercept": bmb.Prior("Normal", mu=125, sigma=3)},
    "log_sigma": {"Intercept": bmb.Prior("Normal", mu=np.log(2), sigma=0.3)},
}

higgs_model = bmb.Model(
    higgs_formula, higgs, family="poisson", link="log", priors=higgs_priors
)
# Spread initial masses across the targeted region; use published background levels
# only as sampler starting values, not as prior centers.
higgs_initial = [
    {
        "b0_category": np.log(higgs_categories["background_per_gev"].to_numpy()),
        "m_h_Intercept": mass,
    }
    for mass in [120, 123, 127, 130]
]
higgs_idata = higgs_model.fit(
    draws=2000,
    tune=1000,
    chains=4,
    cores=2,
    random_seed=123,
    target_accept=0.95,
    init="adapt_diag",
    initvals=higgs_initial,
    progressbar=False,
)
```

We summarize the mass, peak height, and reference width in physical units. We also derive the
**integrated signal yield**, the area under the Gaussian signal summed across categories:

$$
Y_S = A\sqrt{2\pi}\,\sigma\sum_c p_c r_c.
$$

For our 1 GeV bins this is the expected signal event count integrated over the full Gaussian
line shapes; nearly all of that area lies inside the fitted window. Computing it for each
posterior draw retains the dependence between amplitude and width. The intervals below are
95% equal-tailed credible intervals.

```python
def higgs_parameter_draws(idata):
    posterior = idata.posterior.to_dataset()
    parameters = posterior[["m_h_Intercept"]].rename({"m_h_Intercept": "m_h"})
    parameters["amplitude"] = np.exp(posterior["log_amp_Intercept"])
    parameters["sigma_ref"] = np.exp(posterior["log_sigma_Intercept"])
    yield_factor = (
        np.sqrt(2 * np.pi)
        * (
            higgs_categories["peak_fraction"] * higgs_categories["resolution_ratio"]
        ).sum()
    )
    parameters["signal_yield"] = (
        parameters["amplitude"] * parameters["sigma_ref"] * yield_factor
    )
    return parameters


higgs_posterior = higgs_idata.posterior.to_dataset()
higgs_parameters = higgs_parameter_draws(higgs_idata)
az.summary(higgs_parameters, ci_prob=0.95, round_to=3)
```

In physical terms, the posterior mean corresponds to roughly **200 signal events across all
11 categories**, with a 95% credible interval of about 90–350 events. The `amplitude` of about
47 is instead the combined **peak height**: expected signal events in a 1 GeV bin centered on
the resonance. It is not the total number of signal events.

The reference width is about 1.9 GeV. Multiplying it by the fixed category resolution ratios
gives Gaussian widths of roughly **1.1–2.7 GeV** at the posterior mean. These describe how detector
resolution spreads the reconstructed masses, not the Higgs boson's intrinsic decay width.

```python
higgs_diagnostics = az.summary(higgs_idata, kind="diagnostics", round_to=3)
print("Divergences:", int(higgs_idata.sample_stats["diverging"].sum()))
print("Maximum R-hat:", higgs_diagnostics["r_hat"].max())
print("Minimum bulk ESS:", higgs_diagnostics["ess_bulk"].min())
print("Minimum tail ESS:", higgs_diagnostics["ess_tail"].min())
```

## A separately fitted background-only comparison

Removing the resonance changes the best-fitting background. We therefore fit the background-only
model independently, retaining the same data, background parameterization, and background priors.
Its nonlinear parent is simply the log-rate $b_{0c}+b_{1c}x+b_{2c}x^2$. This is a descriptive
comparison of two fits, not a posterior probability that a signal exists.

```python
# | code-fold: true
# | code-summary: "Fit the background-only comparison"
background_only_formula = bmb.Formula(
    "counts ~ b0 + b1 * x + b2 * x**2",
    "b0 ~ 0 + category",
    "b1 ~ 0 + category",
    "b2 ~ 0 + category",
    nlpars=("b0", "b1", "b2"),
)
background_only_model = bmb.Model(
    background_only_formula,
    higgs,
    family="poisson",
    link="log",
    priors={name: higgs_priors[name] for name in ("b0", "b1", "b2")},
)
background_only_idata = background_only_model.fit(
    draws=2000,
    tune=1000,
    chains=4,
    cores=2,
    init="adapt_diag",
    initvals={"b0_category": np.log(higgs_categories["background_per_gev"].to_numpy())},
    random_seed=123,
    target_accept=0.95,
    progressbar=False,
)
background_only_diagnostics = az.summary(
    background_only_idata, kind="diagnostics", round_to=3
)
print("Divergences:", int(background_only_idata.sample_stats["diverging"].sum()))
print("Maximum R-hat:", background_only_diagnostics["r_hat"].max())
print("Minimum bulk ESS:", background_only_diagnostics["ess_bulk"].min())
print("Minimum tail ESS:", background_only_diagnostics["ess_tail"].min())
```

## Displaying the fitted spectrum with category weights

`Model.predict()` evaluates the nonlinear expression on the response scale. For a weighted
display $Y_i=\sum_c w_c N_{ic}$, its expected value is $\sum_c w_c\lambda_{ic}$ and its
variance is $\sum_c w_c^2\lambda_{ic}$. We draw observed error bars using
$\sqrt{\sum_c w_c^2 N_{ic}}$.

Following the CMS prescription, we compute $w_c\propto S_c/(S_c+B_c)$ from the fitted
components within $125\pm\sigma_{\mathrm{eff},c}$, then normalize the weights to preserve
the total fitted signal yield. Our simplified shapes give different numerical weights from
CMS's original fit. These display weights are fixed across posterior draws and never multiply
the likelihood; the calculation is available in the folded code below.

Both panels compare **separately fitted** signal-plus-background and background-only models,
using the same fixed category weights. The lower panel subtracts the background-only posterior
mean from the data and both fitted curves. Its blue curve is the difference between the two
complete fits, **not** the isolated Gaussian signal: the backgrounds can readjust when the signal
is included. This removes the large falling baseline so the local differences are easier to see.

Bands are pointwise 95% credible intervals for expected weighted counts, conditional on the
estimated weights and the fixed subtracted baseline; they are not prediction intervals.
The points' error bars show counting uncertainty only. The gray band shows uncertainty in
the background-only fit relative to that baseline. This visual comparison is not a significance test.

The dotted line marks the [PDG 2026 mass average](https://pdg.lbl.gov/2026/listings/rpp2026-list-higgs-boson.pdf),
$m_H=125.13\pm0.11$ GeV, based on much more data. It is a measured reference, not an exact
true value or an additional constraint on this fit. Both models use 100–180 GeV; the plot
zooms to 110–150 GeV.

Unlike the historical Figure 3 display, our dashed curve is a separately refitted background-only
model. The shaded regions are posterior intervals for expected counts; this figure compares
two fitted models rather than displaying only the background component of an S+B fit.


```python
# | code-fold: true
# | code-summary: "Compute category weights and fitted components"
from scipy.special import ndtr
from scipy.integrate import trapezoid

higgs_fitted = higgs_model.predict(higgs_idata, inplace=False)
# The CSV has category-major rows, then increasing mass within each category.
category_order = higgs["category"].cat.categories
mass_centers = observed_counts.index.to_numpy()
rate = higgs_fitted.posterior["mu"].values.reshape(
    higgs_posterior.sizes["chain"],
    higgs_posterior.sizes["draw"],
    len(category_order),
    -1,
)
b0 = higgs_posterior["b0_category"].values
b1 = higgs_posterior["b1_category"].values
b2 = higgs_posterior["b2_category"].values
background = np.exp(
    b0[..., None]
    + b1[..., None] * ((mass_centers - 125) / 20)
    + b2[..., None] * ((mass_centers - 125) / 20) ** 2
)

# Compute the fixed display weights at posterior mean parameter values.
amplitude = float(higgs_parameters["amplitude"].mean())
mass_mean = float(higgs_parameters["m_h"].mean())
sigma = (
    float(higgs_parameters["sigma_ref"].mean())
    * higgs_categories["resolution_ratio"].to_numpy()
)
peak = amplitude * higgs_categories["peak_fraction"].to_numpy()
half_window = higgs_categories["sigma_eff"].to_numpy()
signal_yield = peak * sigma * np.sqrt(2 * np.pi)
signal_window = signal_yield * (
    ndtr((125 + half_window - mass_mean) / sigma)
    - ndtr((125 - half_window - mass_mean) / sigma)
)
# Integrate the curved background across each category's display-weight window.
window_mass = half_window[:, None] * np.linspace(-1, 1, 101)
window_x = window_mass / 20
background_window = trapezoid(
    np.exp(
        b0.mean(axis=(0, 1))[:, None]
        + b1.mean(axis=(0, 1))[:, None] * window_x
        + b2.mean(axis=(0, 1))[:, None] * window_x**2
    ),
    x=window_mass,
    axis=1,
)
weights = signal_window / (signal_window + background_window)
weights *= signal_yield.sum() / np.sum(weights * signal_yield)

weighted_rate = (rate * weights[None, None, :, None]).sum(axis=2)
background_only_fitted = background_only_model.predict(
    background_only_idata, inplace=False
)
background_only_rate = background_only_fitted.posterior["mu"].values.reshape(rate.shape)
weighted_null = (background_only_rate * weights[None, None, :, None]).sum(axis=2)
assert rate.shape == (4, 2000, 11, 80)
assert np.isfinite(background_only_rate).all() and (background_only_rate > 0).all()
counts_by_category = observed_counts.reindex(columns=category_order).to_numpy().T
weighted_counts = (weights[:, None] * counts_by_category).sum(axis=0)
weighted_errors = np.sqrt((weights[:, None] ** 2 * counts_by_category).sum(axis=0))

# Check that the inverse log link gives background + signal in counts per bin.
expected_signal = (
    np.exp(higgs_posterior["log_amp_Intercept"].values)[..., None, None]
    * higgs_categories["peak_fraction"].to_numpy()[None, None, :, None]
    * np.exp(
        -0.5
        * (
            (mass_centers - higgs_posterior["m_h_Intercept"].values[..., None, None])
            / (
                np.exp(higgs_posterior["log_sigma_Intercept"].values)[..., None, None]
                * higgs_categories["resolution_ratio"].to_numpy()[None, None, :, None]
            )
        )
        ** 2
    )
)
np.testing.assert_allclose(rate, background + expected_signal, rtol=1e-10)
```

```python
# | code-fold: true
# | code-summary: "Compare the fitted models and background-subtracted data"
higgs_fig, (higgs_ax, residual_ax) = plt.subplots(
    2,
    1,
    figsize=(9, 8),
    sharex=True,
    height_ratios=[1, 1.2],
    gridspec_kw={"hspace": 0.08},
)
rate_mean = weighted_rate.mean(axis=(0, 1))
null_mean = weighted_null.mean(axis=(0, 1))
rate_interval = np.quantile(weighted_rate, [0.025, 0.975], axis=(0, 1))
null_interval = np.quantile(weighted_null, [0.025, 0.975], axis=(0, 1))

higgs_ax.errorbar(
    mass_centers,
    weighted_counts,
    yerr=weighted_errors,
    fmt="o",
    color="black",
    markersize=4,
    label="CMS data (weighted)",
)
higgs_ax.plot(
    mass_centers, rate_mean, color=plot_colors[0], label="Signal + background fit"
)
higgs_ax.fill_between(mass_centers, *rate_interval, color=plot_colors[0], alpha=0.25)
higgs_ax.plot(
    mass_centers, null_mean, "--", color=plot_colors[1], label="Background-only fit"
)
higgs_ax.set_ylabel("Weighted events / 1 GeV")
higgs_ax.set_title("CMS diphoton data: fit 100–180 GeV, display 110–150 GeV")

residual_ax.errorbar(
    mass_centers,
    weighted_counts - null_mean,
    yerr=weighted_errors,
    fmt="o",
    color="black",
    markersize=4,
)
residual_ax.fill_between(
    mass_centers,
    *(null_interval - null_mean),
    color="gray",
    alpha=0.2,
    label="Background-only 95% interval",
)
residual_ax.axhline(0, color=plot_colors[1], linestyle="--")
residual_ax.plot(mass_centers, rate_mean - null_mean, color=plot_colors[0])
residual_ax.fill_between(
    mass_centers,
    *(rate_interval - null_mean),
    color=plot_colors[0],
    alpha=0.25,
    label="Signal + background 95% interval",
)
for ax in (higgs_ax, residual_ax):
    ax.axvline(
        125.13,
        color="#735185",
        linestyle=":",
        linewidth=1.5,
        label=r"$m_H$ (PDG 2026): 125.13 GeV",
    )
    ax.set_xlim(110, 150)
higgs_ax.legend(fontsize=10, loc="upper right")
# The reference-line label appears once, in the upper panel.
handles, labels = residual_ax.get_legend_handles_labels()
residual_ax.legend(handles[:2], labels[:2], fontsize=9, loc="upper right")
residual_ax.set_xlabel("Diphoton invariant mass (GeV)")
residual_ax.set_ylabel("Weighted events minus\nbackground-only mean")
plt.show()
```

A shared Gaussian resonance describes the localized structure around the modern Higgs mass
reference, while category-specific backgrounds describe the falling continuum.

The background-only fit leaves a cluster of positive residuals around the modern mass
reference. Adding the shared resonance follows this local structure while allowing the
background to readjust. The counting fluctuations remain substantial: the plot explains what
the signal component contributes, without reproducing the discovery's significance calculation.
Weighting highlights the high-purity categories; both fits still use every original count.

The core nonlinear-model demonstration is now complete: category-specific backgrounds and a
shared resonance have been fitted through one Poisson rate. The following advanced continuation
examines predictive performance and Bayesian model evidence. It is optional for readers focused
on the nonlinear-formula API.

## Advanced: comparing models

### Does the resonance improve prediction?

A more flexible model can follow the observed data better simply because it has more parameters.
**Leave-one-out cross-validation (LOO)** asks a predictive question: how well does each model
predict observations that were left out of its fit? We use Pareto-smoothed importance sampling
(PSIS-LOO) to approximate these held-out predictions from the existing posterior draws, avoiding
hundreds of refits. See [Vehtari et al. (2017)](https://doi.org/10.1007/s11222-016-9696-4)
for the method and its relationship to information criteria.

Here one observation is an **unweighted count in one mass bin and one category**. Both models
use the same 880 observations over 100–180 GeV, background formulas, and background priors.
Other categories' counts at the same mass remain available when a count is held out. This
assesses prediction of missing bin counts within the measured spectrum; it does not hold out
an entire category or mass region, or test extrapolation to a new experiment. The display
weights never enter this calculation.

Bambi computes the pointwise log likelihood through its ordinary model API. ArviZ then estimates
the **expected log predictive density (ELPD)**: larger values mean better held-out prediction.
We show differences relative to background only, with a **paired standard error of the difference**,
not the much larger uncertainty in either model's absolute score. The effective parameter count
summarizes model flexibility; LOO already accounts for its predictive cost.

We also check Pareto-$k$, which diagnoses whether the importance-sampling approximation is
reliable. With 8,000 posterior draws, the recommended threshold is 0.7. Values above it would
require closer investigation or actual held-out refits before trusting the comparison.

```python
higgs_model.compute_log_likelihood(higgs_idata)
background_only_model.compute_log_likelihood(background_only_idata)

higgs_loo = az.loo(higgs_idata, var_name="counts", pointwise=True)
background_only_loo = az.loo(background_only_idata, var_name="counts", pointwise=True)
loo_results = {"Signal + background": higgs_loo, "Background only": background_only_loo}
loo_comparison = az.compare(loo_results, reference="Background only", round_to="none")

for label, result in loo_results.items():
    bad_bins = int((result.pareto_k > result.good_k).sum())
    print(
        f"{label}: max Pareto k = {float(result.pareto_k.max()):.3f}; "
        f"{bad_bins} bins above {result.good_k:.2f}"
    )

loo_comparison[["elpd", "elpd_diff", "dse", "p"]].rename(
    columns={
        "elpd": "ELPD",
        "elpd_diff": "Δ ELPD vs background only",
        "dse": "SE of Δ",
        "p": "Effective parameters",
    }
).round(2)
```

The signal-plus-background model has the higher estimated predictive score, although its
advantage is only about 1.7 standard errors. This is a **modest predictive preference**. All
Pareto-$k$ values are below the threshold, supporting the reliability of the PSIS approximation.

**Why is LOO not our main resonance diagnostic?** The models differ most within a handful of
mass bins. Leaving out one category-bin count retains neighboring bins and the other categories
at the same mass. That is a legitimate test of prediction within the spectrum, but it is a
different question from comparing a coherent localized signal with a background-only hypothesis.

Hundreds of background bins do not automatically dilute this comparison: we compare *paired
pointwise score differences*, and bins predicted similarly by both models largely cancel.
Leaving out an entire category would additionally test prediction for a category whose background
coefficients were not learned. Holding out a whole mass window would test prediction of a peak
whose direct evidence was removed. We retain LOO as a supplementary predictive check.

### From profile likelihood to Bayesian model evidence

In particle physics, a **profile likelihood ratio** compares the best likelihood under
background only with the best likelihood when a signal is allowed. "Profiling" means optimizing
over nuisance parameters, such as each category's background coefficients, under each hypothesis.
It targets the coherent signal across categories directly. See
[Cowan et al. (2011)](https://doi.org/10.1140/epjc/s10052-011-1554-0) for the standard framework;
we will not convert a likelihood ratio into a significance here.

A **Bayes factor** addresses a related model-comparison question by integrating over parameters
instead of optimizing them. For $M_0$ (background only) and $M_1$ (signal plus background),

$$
BF_{10}=\frac{p(y\mid M_1)}{p(y\mid M_0)},\qquad
p(y\mid M_k)=\int p(y\mid\theta_k,M_k)\,p(\theta_k\mid M_k)\,d\theta_k.
$$

The marginal likelihood averages over the specified priors, including background uncertainty
and, under $M_1$, the signal's mass, amplitude, and width. A model that fits well only in a small
part of its prior parameter space pays a corresponding cost. This makes the Bayes factor a useful
but imperfect analogue of a likelihood-ratio comparison: changing a prior can change the Bayes
factor even when the best fit hardly changes. Broadening the possible mass range naturally affects
the evidence; it is not a conversion of a frequentist search correction.

Values above one favor $M_1$. Posterior model odds equal $BF_{10}$ times prior model odds, so a
Bayes factor is not itself the probability that a signal exists. All likelihoods here use the
original Poisson counts, with no display weights.

**Why not just call `az.bayes_factor`?** ArviZ's function uses the Savage–Dickey density ratio at
a specified parameter value. Our background-only model has $A=0$, which is not a finite value
of `log_amp`. Setting `log_amp=0` would test $A=1$, not absence of a signal. We therefore compare
the two separate models using their marginal likelihoods.

We compute approximate, illustrative estimates with [bridge sampling](https://doi.org/10.1016/j.jmp.2017.09.005), using the
Bambi posterior draws and the compiled joint log density of the Bambi-built model, including
all prior and likelihood normalization constants. The folded helper uses a multivariate Normal
proposal fitted to two chains and evaluates the bridge with the other two. It is specialized to
these untransformed Normal coefficients, not a general Bambi evidence API. No custom likelihood
or new dependency is required. We repeat the calculation with reversed chain splits and different
proposal seeds to check numerical stability.


```python
# | code-fold: true
# | code-summary: "Estimate log marginal likelihoods with bridge sampling"
import numpy as np
from scipy.special import logsumexp
from scipy.stats import multivariate_normal


def bridge_log_evidence(
    model, idata, random_seed=123, proposal_draws=8000, reverse=False
):
    """Normal-proposal bridge estimate; only untransformed Normal coefficients here.

    Separate chains fit the proposal and evaluate the bridge (Gronau et al., 2017).
    Returns log marginal likelihood; repeated estimates assess numerical stability.
    """
    backend = model.backend.model
    point = backend.initial_point()
    names = list(point)
    if any(name not in idata.posterior for name in names):
        raise ValueError("This helper requires untransformed posterior variables.")
    if idata.posterior.sizes["chain"] != 4:
        raise ValueError("Use four chains for the two-chain training/evaluation split.")
    shapes = [point[name].shape for name in names]
    sizes = [point[name].size for name in names]
    boundaries = np.cumsum([0] + sizes)
    posterior = np.concatenate(
        [
            idata.posterior[name].values.reshape(
                idata.posterior.sizes["chain"], idata.posterior.sizes["draw"], -1
            )
            for name in names
        ],
        axis=-1,
    )
    if reverse:
        posterior = posterior[::-1]
    split = posterior.shape[0] // 2
    training = posterior[:split].reshape(-1, boundaries[-1])
    evaluation = posterior[split:].reshape(-1, boundaries[-1])
    proposal = multivariate_normal(
        training.mean(axis=0), np.cov(training, rowvar=False)
    )
    candidates = proposal.rvs(
        size=proposal_draws, random_state=np.random.default_rng(random_seed)
    ).reshape(proposal_draws, -1)
    log_joint = backend.compile_logp()

    def log_ratio(draws):
        values = np.array(
            [
                log_joint(
                    {
                        name: row[boundaries[j] : boundaries[j + 1]].reshape(shapes[j])
                        for j, name in enumerate(names)
                    }
                )
                for row in draws
            ]
        )
        return values - proposal.logpdf(draws)

    posterior_ratio = log_ratio(evaluation)
    proposal_ratio = log_ratio(candidates)
    n1, n2 = len(evaluation), len(candidates)
    log_s1, log_s2 = np.log([n1, n2]) - np.log(n1 + n2)
    log_z = np.median(posterior_ratio)
    for iteration in range(1000):
        numerator = proposal_ratio - np.logaddexp(
            log_s1 + proposal_ratio, log_s2 + log_z
        )
        denominator = -np.logaddexp(log_s1 + posterior_ratio, log_s2 + log_z)
        updated = (
            logsumexp(numerator) - np.log(n2) - logsumexp(denominator) + np.log(n1)
        )
        if abs(updated - log_z) < 1e-8:
            return float(updated)
        log_z = updated
    raise RuntimeError("Bridge iteration did not converge")
```

```python
# Four estimates per model: different proposal draws and both chain splits.
bridge_runs = [(123, False), (456, True), (789, False), (1011, True)]


def repeated_log_evidence(model, idata):
    return np.array(
        [
            bridge_log_evidence(model, idata, random_seed=seed, reverse=reverse)
            for seed, reverse in bridge_runs
        ]
    )


log_evidence_background = repeated_log_evidence(
    background_only_model, background_only_idata
)
log_evidence_signal = repeated_log_evidence(higgs_model, higgs_idata)
log_bf = log_evidence_signal - log_evidence_background
print(f"log BF₁₀: {log_bf.mean():.2f}")
rounded_bf = float(f"{np.exp(log_bf.mean()):.2g}")
print(f"BF₁₀: approximately {rounded_bf:g}")
print(f"Repeated log BF₁₀ estimates: {log_bf.min():.2f} to {log_bf.max():.2f}")
```

These are approximate, illustrative evidence estimates. The bridge repeats reuse the same
posterior draws with different proposal seeds and reversed chain splits; they are not independent
posterior runs. They check numerical stability; their range is **not** a credible interval
or a complete error bound. Agreement between repeats cannot rule out posterior modes missed by
all chains. The helper was also checked against a conjugate Normal model with an analytically
known marginal likelihood, and the Higgs fits were checked from dispersed mass starting points. In the wider-mass-prior
fit, a rare off-peak tail is visited unevenly across chains: 56 of 8,000 draws fall outside
120–130 GeV, all in one chain. Good R-hat values alone do not establish complete exploration
of that tail. We therefore report rounded evidence estimates and keep this numerical limitation
separate from sensitivity to the modeling assumptions. A separate 16,000-draw check with a new
seed and dispersed mass and amplitude starting values gave consistent log marginal likelihoods
(about -2467.25 to -2467.23 for the wider-prior S+B model). Its rare off-peak visits were also
uneven across chains: agreement supports the rounded result without removing the tail-exploration
limitation.

The Bayes factor favors S+B under the stated priors. To see how conditional that conclusion is,
we refit with a wider mass prior, Normal$(125,10)$, or a wider log-amplitude prior,
Normal$(\log40,1.2)$, changing one at a time. The background and width priors remain fixed.
This is a targeted sensitivity check, not a reconstruction of the historical search prior.

```python
# | code-fold: true
# | code-summary: "Refit under two alternative signal priors"
evidence_variants = [
    ("Wider mass prior", "m_h", bmb.Prior("Normal", mu=125, sigma=10)),
    ("Wider amplitude prior", "log_amp", bmb.Prior("Normal", mu=np.log(40), sigma=1.2)),
]
evidence_sensitivity = {"Targeted priors": log_bf}
for label, parameter, prior in evidence_variants:
    variant_model = bmb.Model(
        higgs_formula,
        higgs,
        family="poisson",
        link="log",
        priors=higgs_priors | {parameter: {"Intercept": prior}},
    )
    initial_masses = (
        [115, 125, 135, 145] if parameter == "m_h" else [120, 123, 127, 130]
    )
    variant_initial = [
        {
            "b0_category": np.log(higgs_categories["background_per_gev"].to_numpy()),
            "m_h_Intercept": mass,
        }
        for mass in initial_masses
    ]
    variant_idata = variant_model.fit(
        draws=2000,
        tune=1000,
        chains=4,
        cores=2,
        init="adapt_diag",
        initvals=variant_initial,
        random_seed=123,
        target_accept=0.95,
        progressbar=False,
    )
    diagnostics = az.summary(variant_idata, kind="diagnostics", round_to=3)
    print(
        f"{label}: divergences={int(variant_idata.sample_stats['diverging'].sum())}, "
        f"max R-hat={diagnostics['r_hat'].max():.3f}, "
        f"min bulk ESS={diagnostics['ess_bulk'].min():.0f}, "
        f"min tail ESS={diagnostics['ess_tail'].min():.0f}"
    )
    if parameter == "m_h":
        mass_draws = variant_idata.posterior["m_h_Intercept"]
        off_peak = ((mass_draws < 120) | (mass_draws > 130)).sum("draw")
        print("Draws outside 120–130 GeV, by chain:", off_peak.values)
    evidence_sensitivity[label] = (
        repeated_log_evidence(variant_model, variant_idata) - log_evidence_background
    )
```

```python
# | code-fold: true
# | code-summary: "Format the evidence comparison table"
evidence_rows = []
for label, values in evidence_sensitivity.items():
    rounded_bf = float(f"{np.exp(values.mean()):.2g}")
    evidence_rows.append(
        {
            "Signal priors": label,
            "log BF₁₀": values.mean(),
            "Approx. BF₁₀": f"{rounded_bf:g}",
            "Repeat minimum (log BF)": values.min(),
            "Repeat maximum (log BF)": values.max(),
        }
    )
pd.DataFrame(evidence_rows).set_index("Signal priors").round(2)
```

The approximate Bayes factors strongly favor S+B over this background-only alternative under
the three signal-prior specifications examined. Their magnitude changes materially: a wider mass
prior spreads probability over locations where the data do not support a resonance, and a wider
amplitude prior spreads probability over very weak and very large signals. That dependence is
part of the question a Bayes factor answers, not just numerical uncertainty.

**These checks vary signal priors within one fixed background family.** They do not establish
robustness of the Bayes factor to background modeling. The comparison below changes the inferred
signal yield from about 204 to 90 events when background curvature is removed, illustrating why
this qualification matters.

These are **retrospective comparisons**, conditional on modeling choices informed by the observed
spectrum and published Higgs information. We selected a curved background after examining the
spectrum and used a known-region signal prior. The Bayes factors integrate over parameters within
the final specified models; they do not account for that model-development process. Broad, proper
background priors do not remove this limitation.

LOO and the Bayes factor need not express the same strength of preference: LOO assesses
held-out prediction from fitted posteriors, whereas the Bayes factor compares the models' prior
predictive probabilities of the whole observed dataset. Neither result establishes that a Gaussian
resonance model is literally true or reproduces the original Higgs discovery analysis.

### How sensitive is the signal to the background?

The full 100–180 GeV interval provides more background information, but a wider range also
requires more flexibility in its shape. As a sensitivity check, we remove `b2` and fit a
single exponential in each category, using the same full-range data, signal, and other priors.
This is an illustrative comparison, not CMS's polynomial selection procedure or a validation
of unbiased signal extraction. The additional fit is folded below.

```python
# | code-fold: true
# | code-summary: "Check sensitivity to removing background curvature"
exponential_formula = bmb.Formula(
    "counts ~ log(exp(b0 + b1 * x) + "
    "peak_fraction * exp(log_amp - 0.5 * "
    "((mass - m_h) / (exp(log_sigma) * resolution_ratio))**2))",
    "b0 ~ 0 + category",
    "b1 ~ 0 + category",
    nlpars=("b0", "b1", "log_amp", "m_h", "log_sigma"),
)
exponential_model = bmb.Model(
    exponential_formula,
    higgs,
    family="poisson",
    link="log",
    priors={name: prior for name, prior in higgs_priors.items() if name != "b2"},
)
exponential_idata = exponential_model.fit(
    draws=1000,
    tune=1000,
    chains=4,
    cores=2,
    init="adapt_diag",
    initvals=higgs_initial,
    random_seed=123,
    target_accept=0.95,
    progressbar=False,
)
exponential_parameters = higgs_parameter_draws(exponential_idata)
exponential_diagnostics = az.summary(exponential_idata, kind="diagnostics", round_to=3)
print("Divergences:", int(exponential_idata.sample_stats["diverging"].sum()))
print("Maximum R-hat:", exponential_diagnostics["r_hat"].max())
print("Minimum bulk ESS:", exponential_diagnostics["ess_bulk"].min())
print("Minimum tail ESS:", exponential_diagnostics["ess_tail"].min())
```

```python
# | code-fold: true
# | code-summary: "Compare posterior means and 95% credible intervals"
comparison_rows = []
for label, parameters in [
    ("Curved log-rate (primary)", higgs_parameters),
    ("Exponential", exponential_parameters),
]:
    row = {"Background": label}
    for parameter, column in [
        ("m_h", "Peak mass (GeV)"),
        ("signal_yield", "Signal yield (events)"),
    ]:
        values = parameters[parameter]
        lower, upper = np.quantile(values, [0.025, 0.975])
        row[column] = f"{float(values.mean()):.2f} [{lower:.2f}, {upper:.2f}]"
    comparison_rows.append(row)
pd.DataFrame(comparison_rows).set_index("Background")
```

The peak location remains close to 125 GeV under both backgrounds, while the estimated signal
yield changes more noticeably. The single exponential pulls the inferred signal yield down substantially over this wider
range. This is a warning that background assumptions can matter even when the peak location
appears stable; the comparison does not establish that the curved background is unbiased. Recovering the
mass peak is easier than reproducing a precision signal-strength measurement; the latter
requires dedicated background-bias studies and calibrated signal shapes.

The central lesson is composition: a background and a resonance combine in one Poisson rate.
Bambi's ordinary parameter formulas let the background vary by category, while the nonlinear
expression estimates a shared peak location, amplitude, and width. The standard Bambi fitting and prediction APIs handle the combined model.
